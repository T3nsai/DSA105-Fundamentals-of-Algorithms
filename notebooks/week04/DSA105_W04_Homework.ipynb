# DSA105 TRACE Homework — Week 4

A service sells:

- Standard plan: 85 subscriptions × 350 TL
- Premium plan: 40 subscriptions × 390 TL

Calculate:

- each plan's revenue;
- total revenue;
- premium revenue share;
- average revenue per subscription.

Write readable expressions with meaningful result names.


```python
standard_subscriptions = 85
standard_price = 350
premium_subscriptions = 40
premium_price = 390

# Complete the calculations.

standard_plan_revene = standard_subscriptions * standard_price
premium_plan_revenue = premium_subscriptions * premium_price
total_revenue = standard_plan_revene + premium_plan_revenue
premium_revenue_share = premium_plan_revenue / total_revenue
avarage_revenue_per_subscription = total_revenue / (standard_subscriptions + premium_subscriptions)

print(f"Standard plan revenue: {standard_plan_revene} TL")
print(f"Premium plan revenue: {premium_plan_revenue} TL")
print(f"Total revenue: {total_revenue} TL")
print(f"Premium revenue share: {premium_revenue_share}")
print(f"Avarage revenue per subscription: {avarage_revenue_per_subscription} TL")
```
