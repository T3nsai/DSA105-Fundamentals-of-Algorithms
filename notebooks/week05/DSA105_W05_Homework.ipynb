Build an interactive delivery calculator.

Ask for:

- distance in kilometers;
- price per kilometer.

Calculate the delivery cost and display it with two decimal places.


```python
# Input and conversion
distance = float(input("Distance in kilometers: "))
price = float(input("Price per kilometer: "))

# Computation
delivery_cost = distance * price

# Output
print(f"Delivery cost: {delivery_cost:.2f}")

```
