### 2. Convert before calculating


```python
quantity = int(input("Quantity: "))
unit_price = float(input("Unit price: "))

total_price = quantity * unit_price

print(f"Total price: {total_price:.2f}")
```

Read the program as three blocks:

1. input and conversion;
2. computation;
3. output.

The blank lines make that structure visible without adding any new Python rule.

### 5. Lab — event registration

Ask the user for:

- number of participants;
- fee per participant.

Calculate and display the total fee.


```python
# Input and conversion
number_of_participants = int(input("Number of participants: "))
fee_per_participant = float(input("Fee per participant: "))

# Computation
total_cost = number_of_participants * fee_per_participant

# Output
print(f"Total cost: {total_cost:.2f}")
```
