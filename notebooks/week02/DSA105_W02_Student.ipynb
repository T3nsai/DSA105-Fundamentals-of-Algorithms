### 7. Independent build — Lab Evidence

A delivery service completed:

- 18 standard deliveries at 65 TL;
- 9 express deliveries at 110 TL.

Build a sequential program that calculates:

- standard revenue;
- express revenue;
- total revenue.

Predict the final total before running.


````python
standard_delivery_price = 65
standard_delivery_count = 18
express_delivery_price = 110
express_delivery_count = 9

# Build the remaining statements here.

standard_revenue = standard_delivery_price * standard_delivery_count
express_revenue = express_delivery_price * express_delivery_count
total_revenue = standard_revenue + express_revenue

print("Standard revenue:",standard_revenue)
print("Express revenue:", express_revenue)
print("Total revenue",total_revenue)
````
